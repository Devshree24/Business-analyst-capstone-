```python
# comparing more than two conditions  mood vs genre using Kruskal-Wallis
comapring Mood_After_Listening across Favorite_Music_Genre categories.
```

```python
import pandas as pd
from scipy.stats import kruskal
```

```python
# Interpretation
if p_value < 0.05:
    print("Conclusion: There is a statistically significant difference in mood after listening among different music genres.")
else:
    print("Conclusion: There is NO statistically significant difference in mood after listening among different music genres.")
```

```python
# using one way anova test To test whether there's a significant difference in hours of listening between different music genres
```

```python
import pandas as pd
from scipy.stats import f_oneway
```

```python
df = df.dropna(subset=['Favorite_Music_Genre', 'Hours_Listening_Per_Day'])
```

```python
genre_groups = df.groupby('Favorite_Music_Genre')['Hours_Listening_Per_Day'].apply(list)
```

```python
# Perform one-way ANOVA
f_stat, p_value = f_oneway(*genre_groups)
```

```python
print("One-way ANOVA Result: Hours of Listening by Music Genre")
print(f"F-statistic = {f_stat:.3f}")
print(f"p-value = {p_value:.3f}")
```

```python
# Interpretation
if p_value < 0.05:
    print("Conclusion: There is a statistically significant difference in listening hours among music genres.")
else:
    print("Conclusion: There is NO statistically significant difference in listening hours among music genres.")
```

```python
# using chi square test to associate betwwen the mental health condition and favourite music genere 
```

```python
import pandas as pd
from scipy.stats import chi2_contingency
```

```python
df = df.dropna(subset=['Favorite_Music_Genre', 'Mental_Health_Condition'])
```

```python
contingency_table = pd.crosstab(df['Favorite_Music_Genre'], df['Mental_Health_Condition'])
```

```python
# Perform Chi-Square Test
chi2_stat, p_value, dof, expected = chi2_contingency(contingency_table)
```

```python
print("Chi-Square Test: Association between Music Genre and Mental Health Condition")
print(f"Chi-Square Statistic = {chi2_stat:.3f}")
print(f"Degrees of Freedom = {dof}")
print(f"p-value = {p_value:.3f}")
```

```python
#interpretation
if p_value < 0.05:
    print("Conclusion: There is a statistically significant association between favorite music genre and mental health condition.")
else:
    print("Conclusion: There is NO statistically significant association between favorite music genre and mental health condition.")
```

```python
# importing the libraries for finding the MAPE, COPE,and Durbin watson value and while predicting the model of hours of listening based on age and gender 
```

```python
import pandas as pd
import numpy as np
import statsmodels.api as sm
from sklearn.preprocessing import LabelEncoder
from sklearn.metrics import mean_absolute_percentage_error
from statsmodels.stats.stattools import durbin_watson
```

```python
df = df.dropna(subset=['Age', 'Gender', 'Hours_Listening_Per_Day'])
```

```python
label_encoder = LabelEncoder()
df['Gender_Encoded'] = label_encoder.fit_transform(df['Gender'])
```

```python
X = df[['Age', 'Gender_Encoded']]
y = df['Hours_Listening_Per_Day']
```

```python
X_const = sm.add_constant(X)
```

```python
model = sm.OLS(y, X_const).fit()
```

```python
y_pred = model.predict(X_const)
```

```python
# Calculate MAPE
mape = mean_absolute_percentage_error(y, y_pred)
```

```python
# Calculate COPE = 1 - (SSE / SST)
sse = np.sum((y - y_pred)**2)
sst = np.sum((y - np.mean(y))**2)
cope = 1 - (sse / sst)
```

```python
# Durbin-Watson statistic
dw_stat = durbin_watson(y - y_pred)
```

```python
print("Linear Regression Model Summary:")
print(model.summary())
print("\nEvaluation Metrics:")
print(f"MAPE: {mape:.3f}")
print(f"COPE (Coefficient of Prediction Efficiency): {cope:.3f}")
print(f"Durbin-Watson Statistic: {dw_stat:.3f}")
```

```python
Durbin-Watson: Checks for autocorrelation in residuals. Values:
≈ 2 → No autocorrelation
< 2 → Positive autocorrelation
2 → Negative autocorrelation
```

```python
# Perform paired t-test
import pandas as pd
from scipy.stats import ttest_rel
```

```python
df = pd.read_csv("C:/Users/Devshree Gupta/Downloads/Music DS w2.csv")
```

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from scipy.stats import ttest_rel
```

```python
df.columns = df.columns.str.strip()
df['Mood_Before_Listening'] = pd.to_numeric(df['Mood_Before_Listening'], errors='coerce')
df['Mood_After_Listening'] = pd.to_numeric(df['Mood_After_Listening'], errors='coerce')
df_clean = df.dropna(subset=['Mood_Before_Listening', 'Mood_After_Listening'])
```

```python
df_clean['Mood_Change'] = df_clean['Mood_After_Listening'] - df_clean['Mood_Before_Listening']
```

```python
print("Descriptive Statistics:")
print(df_clean[['Mood_Before_Listening', 'Mood_After_Listening', 'Mood_Change']].describe())
```

```python
t_stat, p_value = ttest_rel(df_clean['Mood_Before_Listening'], df_clean['Mood_After_Listening'])
```

```python
print("\nPaired t-test Result:")
print(f"t-statistic = {t_stat:.3f}")
print(f"p-value     = {p_value:.3f}")
```

```python
if p_value < 0.05:
    print("Statistically significant mood improvement after listening to music.")
else:
    print("No statistically significant change in mood detected after listening to music.")
```

```python
# Boxplot Visualization
plt.figure(figsize=(8, 5))
sns.boxplot(data=df_clean[['Mood_Before_Listening', 'Mood_After_Listening']])
plt.title("Mood Before vs After Listening to Music")
plt.ylabel("Mood Score")
plt.show()
```
